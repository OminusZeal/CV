```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread(r"C:\Users\JAI\Downloads\lenna.jpg",1)
```

```python
img.shape
```

```python
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
RGB_img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.subplot(1,2,1)
plt.imshow(RGB_img)
plt.axis('off')
plt.title("RGB")

gray=cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
plt.subplot(1,2,2)
plt.imshow(gray,cmap='gray')
plt.axis('off')
plt.title("GRAY")
```

```python
equ = cv2.equalizeHist(gray)
res = np.hstack((gray, equ))
plt.axis('off')
plt.imshow(res,cmap='gray')
```

```python
plt.subplot(1, 2, 1)
plt.hist(img.ravel(), 256, [0, 256], color='gray')
plt.title('Original Image Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(equ.ravel(), 256, [0, 256], color='gray')
plt.title('Equalized Image Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
height, width, channels = img.shape
gray_img = np.zeros((height, width), dtype=np.uint8)

for i in range(height):
    for j in range(width):
        b, g, r = img[i, j]
        gray = int(0.299 * r + 0.587 * g + 0.114 * b)
        gray_img[i, j] = gray

cv2.imshow('Original Image', img)
cv2.imshow('Grayscale Image', gray_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img = cv2.imread(r"C:\Users\JAI\Downloads\lenna.jpg",0)
hist = np.zeros(256, dtype=int)
for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        hist[img[i, j]] += 1

cdf = np.cumsum(hist)
cdf_normalized = cdf / cdf[-1]

equalized_img = np.zeros_like(img)
for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        equalized_img[i, j] = int(cdf_normalized[img[i, j]] * 255)

cv2.imshow('Original Image', img)
cv2.imshow('Equalized Image', equalized_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plt.subplot(1, 2, 1)
plt.hist(img.ravel(), 256, [0, 256], color='gray')
plt.title('Original Image Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(equalized_img.ravel(), 256, [0, 256], color='gray')
plt.title('Equalized Image Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```
