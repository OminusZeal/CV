```python
import cv2

# Load video
cap = cv2.VideoCapture(r"C:\Users\JAI\Downloads\vdeo.mp4")

# Create background subtractor
fgbg = cv2.createBackgroundSubtractorMOG2(history=500,varThreshold=10,detectShadows=True)

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    
    # Apply background subtraction
    fgmask = fgbg.apply(frame)

    # Display results
    cv2.imshow("Original", frame)
    cv2.imshow("Foreground Mask", fgmask)

    if cv2.waitKey(30) & 0xFF == 27:  # Press ESC to exit
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Load video
cap = cv2.VideoCapture(r"C:\Users\JAI\Downloads\vdeo.mp4")

# Read first frame
ret, frame = cap.read()
avg_frame = np.float32(frame)

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Compute weighted average
    cv2.accumulateWeighted(frame, avg_frame, alpha=0.05)

    # Convert to displayable format
    avg_display = cv2.convertScaleAbs(avg_frame)

    # Display
    cv2.imshow("Original", frame)
    cv2.imshow("Averaged Frame", avg_display)

    if cv2.waitKey(30) & 0xFF == 27:  # Press ESC to exit
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Load video
cap = cv2.VideoCapture(r"C:\Users\JAI\Downloads\vdeo.mp4")

# Read first frame
ret, prev_frame = cap.read()
prev_gray = cv2.cvtColor(prev_frame, cv2.COLOR_BGR2GRAY)

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Convert to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Compute absolute difference
    diff = cv2.absdiff(prev_gray, gray)

    # Thresholding to highlight differences
    _, thresh = cv2.threshold(diff, 30, 255, cv2.THRESH_BINARY)

    # Display
    cv2.imshow("Original", frame)
    cv2.imshow("Scene Change Detection", thresh)

    # Update previous frame
    prev_gray = gray.copy()

    if cv2.waitKey(30) & 0xFF == 27:  # Press ESC to exit
        break

cap.release()
cv2.destroyAllWindows()
```
